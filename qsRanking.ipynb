```python
import pandas as pd
```

```python
df = pd.read_csv('qs-world-rankings-2025.csv')
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop(df.index[:500], inplace=True)
```

```python
df['2025 Rank'] = df['2025 Rank'].str.extract(r'(\d{3})')
df['2024 Rank'] = df['2024 Rank'].str.extract(r'(\d{3})')
```

```python
df['2025 Rank'] = pd.to_numeric(df['2025 Rank'].str.replace("+", " "))
df['2024 Rank'] = pd.to_numeric(df['2024 Rank'].str.replace("+", " "))
df['QS Overall Score'] = pd.to_numeric(df['QS Overall Score'].str.replace(",", " "))
```
